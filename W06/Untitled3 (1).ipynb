```python
# 匯入資料科學的核心套件
import numpy as np
import pandas as pd

# 設定亂數種子以確保結果可重現
random_seed = 42
rng = np.random.default_rng(random_seed)

print("套件匯入成功！亂數產生器已準備就緒。")
```

```python
student_count = 60
n = student_count

# 1. 學生編號 (1 到 60)
student_id = np.arange(1, n + 1)

# 2. 隨機抽取科系
departments_list = ["資訊管理", "企業管理", "金融管理"]
department = rng.choice(departments_list, size=n)

# 3. 睡眠時數 (平均 6.7, 標準差 1.1, 四捨五入至小數點第 1 位)
sleep_hours = np.round(rng.normal(loc=6.7, scale=1.1, size=n), 1)

# 4. 螢幕使用時間 (平均 5.0, 標準差 1.5, 限制在 1~10 小時)
screen_time_hours = np.round(
    np.clip(rng.normal(loc=5.0, scale=1.5, size=n), 1.0, 10.0), 1
)

# 5. 讀書時數 (平均 2.5, 標準差 1.0, 限制在 0.5~6 小時)
study_hours = np.round(
    np.clip(rng.normal(loc=2.5, scale=1.0, size=n), 0.5, 6.0), 1
)

# 6. 出席率 (平均 88, 標準差 8, 限制在 60~100)
attendance_rate = np.round(
    np.clip(rng.normal(loc=88, scale=8, size=n), 60, 100)
)

# 組合成 DataFrame
df = pd.DataFrame({
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
})

# 顯示前五筆資料
df.head()

```

```python
# 產生平均為 0、標準差為 5 的隨機誤差
random_error = rng.normal(loc=0, scale=5, size=n)

# 套用公式計算小考成績
raw_quiz_score = (
    50
    + (3 * df["study_hours"])
    + (2 * df["sleep_hours"])
    - (1.2 * df["screen_time_hours"])
    + (0.15 * df["attendance_rate"])
    + random_error
)

# 將成績限制在 0~100 分並四捨五入成整數
df["quiz_score"] = np.round(np.clip(raw_quiz_score, 0, 100)).astype(int)

# 使用多重選取，顯示特定欄位的前五筆資料
selected_columns = [
    "student_id",
    "study_hours",
    "sleep_hours",
    "screen_time_hours",
    "attendance_rate",
    "quiz_score"
]

df[selected_columns].head()
```
