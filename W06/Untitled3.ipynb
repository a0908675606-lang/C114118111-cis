```python
# 第四單元：算術運算與分數計算
# 設有一位學生的學習生活資料
study_hours = 2.5
sleep_hours = 7.0
screen_time_hours = 4.0
attendance_rate = 90

# 分步計算各項影響分數
base_score = 50
study_effect = 3 * study_hours         # 讀書時間加分
sleep_effect = 2 * sleep_hours         # 睡眠時間加分
screen_effect = -1.2 * screen_time_hours # 螢幕時間扣分
attendance_effect = 0.15 * attendance_rate # 出席率加分

# 總分合計
final_score = base_score + study_effect + sleep_effect + screen_effect + attendance_effect

print("--- 影響分數明細 ---")
print("讀書加分：", study_effect)
print("睡眠加分：", sleep_effect)
print("螢幕扣分：", screen_effect)
print("出席加分：", attendance_effect)
print("--------------------")
print("預估小考成績：", final_score)
```

```python
# 第五單元：定義分數評級函式
def describe_score(score):
    # 判斷分數等級
    if score >= 80:
        return "表現良好"
    elif score >= 60:
        return "達到基本標準"
    else:
        return "需要加強"

# 測試函式呼叫
score_1 = 85.5
score_2 = 72.0
score_3 = 45.0

print(f"分數 {score_1} 的評級為：", describe_score(score_1))
print(f"分數 {score_2} 的評級為：", describe_score(score_2))
print(f"分數 {score_3} 的評級為：", describe_score(score_3))
```

```python
# 第七單元：建立 60 筆模擬學生資料
student_count = 60
n = student_count

# 1. 學生編號 (1 到 60)
student_id = np.arange(1, n + 1)

# 2. 隨機抽取科系
departments_list = ["資訊管理", "企業管理", "金融管理"]
department = rng.choice(departments_list, size=n)

# 3. 睡眠時數（平均 6.7，標準差 1.1）
sleep_hours = rng.normal(loc=6.7, scale=1.1, size=n)
sleep_hours = np.round(sleep_hours, 1)

# 4. 螢幕使用時數（平均 5.0，標準差 1.5，限制在 1~10 小時）
screen_time_hours = rng.normal(loc=5.0, scale=1.5, size=n)
screen_time_hours = np.clip(screen_time_hours, 1.0, 10.0)
screen_time_hours = np.round(screen_time_hours, 1)

# 5. 讀書時數（平均 2.5，標準差 1.0，限制在 0.5~6 小時）
study_hours = rng.normal(loc=2.5, scale=1.0, size=n)
study_hours = np.clip(study_hours, 0.5, 6.0)
study_hours = np.round(study_hours, 1)

# 6. 出席率（平均 88，標準差 8，限制在 60~100）
attendance_rate = rng.normal(loc=88, scale=8, size=n)
attendance_rate = np.clip(attendance_rate, 60, 100)
attendance_rate = np.round(attendance_rate, 0)

# 組合成為字典並轉換為 DataFrame
data = {
    "student_id": student_id,
    "department": department,
    "sleep_hours": sleep_hours,
    "screen_time_hours": screen_time_hours,
    "study_hours": study_hours,
    "attendance_rate": attendance_rate
}

df = pd.DataFrame(data)

# 顯示前五筆資料
df.head()
```
